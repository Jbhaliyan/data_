```python
from pyspark.sql.types import StructType ,StructField, IntegerType,StringType,DecimalType,ArrayType
from pyspark.sql import functions as F

# from pysaprk.sql.array 
```

```python
# Sample data
data = [
    (1, "Alice", ["Math", "Science", "English"], [85, 90, 78]),
    (2, "Bob", ["Math", "Science", "English"], [88, 76, 92]),
    (3, "Charlie", ["Math", "Science", "English"], [65, 70, 80])
]

# Define schema
schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("subjects", ArrayType(StringType()), True),
    StructField("subject_marks", ArrayType(IntegerType()), True)
])

# Create DataFrame
df = spark.createDataFrame(data, schema)

# Show the DataFrame
df.show(truncate=False)
```

##to expand the above df so it shows marks for each subject in seperate rows

```python
subjects = df.select("subjects").collect()[0][0]
print(subjects)

df_1 = df.withColumn("mapping", F.arrays_zip(F.col("subjects"),F.col("subject_marks")))
df_1.display()
df_2 =df_1.select("id","name",F.explode(F.col("mapping")))
df_2.display()
df_3 =df_2.select("id","name","col.subjects","col.subject_marks")
df_3.display()

# df_2 = df_1.withColumn("new", F.flatten(F.col("mapping")))
# df_2.display()
```
