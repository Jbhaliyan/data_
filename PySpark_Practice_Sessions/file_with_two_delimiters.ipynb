```python
file_path = "/Volumes/workspace/practice/practice_data"
file_name = "Data.csv"
```

```python
# delimeter option is not working here
df = spark.read.option("header","true")\
    .option("delimeter","~|")\
    .option("ignore",",")\
    .csv(file_path + "/" + file_name)
df.display()
```

```python
# since the file is stored as .csv and we have two delimiters along with "," in the name and not as delimiter so we read it as text

df = spark.read.text(file_path + "/" + file_name)
df.display()
header = df.first()
schema = header[0].split("~|")

print(f"First row from dataframe: {header}\nSchema from header row: {schema}")
```

```python
# to remove header row from dataframe
df_new = df.filter(df.value != header[0])
df_new.display()
# since there couls be n number of columns and we can't keep creating columns individually so we convert dataframe to rdd 
df1  = df_new.rdd


# df1 = df_new.withColumn("Name", F.split(F.col("value"),"[~|]")[0])\
#     .withColumn("Age", F.split(F.col("value"),"[~|]")[2]).drop("value")
# # df2 =df1.withColumn("Name", F.col("cols")[0])\
# #     .withColumn("Age", F.col("cols")[2]).drop("cols")
# df1.display()
```
