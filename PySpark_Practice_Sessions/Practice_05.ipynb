####For the data given calculate KPI column as sales/avg(last 2 weeks)-1 for each merchant per week.

```python
from pyspark.sql import functions as F
```

```python
data = [
    ('01','Merchant1',24),
    ('01','Merchant2',68),
    ('02','Merchant1',45),
    ('02','Merchant2',79),
    ('03','Merchant1',90),
    ('04','Merchant1',23)
]
columns = ['Week','Merchant','Sales']
sales = spark.createDataFrame(data,columns)
sales.display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.types import DecimalType
# NOTE: to look back for 2 previous weeks for every merchant
window = Window.partitionBy("Merchant").orderBy("Week").rowsBetween(-2,-1)
df_1 = sales.withColumn("sales_in_lst_2_wks",F.collect_list("Sales").over(window))\
    .withColumn("avg_sal_lst_2_wk",F.avg('Sales').over(window))\
    .withColumn("KPI", (F.col('Sales')/ (F.col('avg_sal_lst_2_wk')-1)).cast(DecimalType(10,2)))
df_1.display()
```
