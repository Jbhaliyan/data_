```python
# NOTE: this initialization is unnecessary in databricks
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('Practice_02').getOrCreate()
```

```python
from pyspark.sql.types import StructType,StructField
```

```python
data = [
    (101, "Alice", "Math", 85),
    (101, "Alice", "Chemistry", 85),
    (101, "Alice", "Physics", 78),
    (102, "Bob", "Math", 90),
    (102, "Bob", "Physics", 88),
    (103, "Charlie", "Math", 75),
    (103, "Charlie", "Chemistry", 75),
    (104, "David", "Math", 88),
    (104, "David", "Chemistry", 88),
    (104, "David", "Physics", 95),
    (105, "Eve", "Chemistry", 91),
]
columns = ["Student_ID", "Name", "Subject", "Marks"]
student = spark.createDataFrame(data,columns)
student.display()
```

Find name and ID of student with same marks in Math and Chemistry

```python
student_1 = student.filter("Subject in ('Math','Chemistry')")
student_1.display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import lag
window = Window.partitionBy("Student_ID").orderBy("Subject")
student_2 = student_1.withColumn("prev_marks", lag("Marks",1).over(window))
student_2.display()
```

```python
student_2.select("Student_ID","Name").filter("prev_marks is not null").display()
```
