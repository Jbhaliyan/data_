```python
from pyspark.sql import functions as F
```

```python
data = [("John,Smith", "Canada"), ("Mike,David", "USA")]
df = spark.createDataFrame(data,["Names","Country"])
df.display()
```

```python
df1 = df.withColumn("name_lst",F.split(F.col("Names"),","))
df2 = df1.withColumn("Names", F.explode(F.col("name_lst"))).drop("name_lst")
df2.display()
```
