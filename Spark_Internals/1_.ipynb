```python
df = spark.table("movies")
df.display()
```

```python
display(df.summary())
```

### Column Filtering


```python
df.select('title','imdb_rating','language').show(3,truncate=False)
# truncate =false to display the entire content of long lenght string column s such as title in above case.
```

### Row Filtering

```python
display(df.filter((df['release_year'] >= 2000) & (df['release_year'] <= 2010)))
```

```python
# second way to filter movies released between year 2000 and 2010 
from pyspark.sql.functions import col

df_1 = df.filter( (col("release_year")>=2000) & 
                  (col("release_year")<=2010)
                  )
df_1.display()
```

```python
df.filter(col("release_year").between(2000,2010)).display()
```
