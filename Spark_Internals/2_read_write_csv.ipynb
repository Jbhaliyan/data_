```python
from pyspark.sql import functions as f, types as t

csv_schema = t.StructType([
    t.StructField("order_date", t.DateType()),
    t.StructField("country", t.StringType()),
    t.StructField("order_id", t.IntegerType()),
    t.StructField("product", t.StringType()),
    t.StructField("qty", t.IntegerType()),
    t.StructField("price", t.DoubleType())
])

df = spark.read \
    .schema(csv_schema)\
    .option("header",True).csv("/Volumes/workspace/default/raw_data/orders.csv")
df.display()
```

```python
# To fix the schema of columns 
df = spark.read \
    .option("inferSchema",True)\
    .option("dateFormat","MM-dd-yyyy")\
    .option("header",True).csv("/Volumes/workspace/default/raw_data/orders.csv")
df.display()
```

```python
# ANother way to load CSV file
df = spark.read \
    .format("csv") \
    .option("inferSchema", True) \
    .option("header", True) \
    .load("/Volumes/workspace/default/raw_data/orders.csv")
df.display()
```

```python
# Write file in parquet format

df.write.mode("overwrite").parquet("/Volumes/workspace/default/raw_data/orders")
```
